Text-to-Speech Audio Generation
Modern applications increasingly require high-quality audio generation for purposes such as audiobooks, e-learning content, accessibility tools, voice assistants, video narration, and interactive applications. However, generating natural, human-sounding speech traditionally requires complex software, licensed voices, or expensive production tools.
To simplify this process, we want to build a fully automated Text-to-Speech (TTS) system using OpenAI’s TTS-1-HD model inside Google Colab. This system should allow users to input any text and receive a high-quality MP3 audio file generated by the AI.
Objectives of the System
1.	Convert user-provided text into natural-sounding speech using OpenAI’s API.
2.	Allow selection of high-quality voices such as nova, alloy, echo, fable, shimmer, etc.
3.	Generate HD-quality audio suitable for:o	Audiobooks
	Presentations
	Podcasts
	Educational narration
	Accessibility for visually impaired users
4.	Automatically save the generated audio file in MP3 format.
5.	Enable playback directly inside Google Colab.
6.	Allow the user to easily download the audio file for external use.

```python
# Install OpenAI library
!pip install openai

# Import required libraries
from openai import OpenAI
from google.colab import userdata
from IPython.display import Audio, display


api_key = userdata.get('OPENAI_API_KEY')

# Initialize the OpenAI client
client = OpenAI(api_key=api_key)

# Text to convert to speech
text = """
Hello! This is a high-quality text-to-speech example using OpenAI's API in Google Colab.
You can use this for various applications like audiobooks, presentations,
or educational content. The audio quality is optimized for the best listening experience.
"""

# Generate speech with high quality settings
print("Generating audio...")
response = client.audio.speech.create(
    model="tts-1-hd",  # High-definition model for better quality
    voice="nova",       # Options: alloy, echo, fable, onyx, nova, shimmer
    input=text,
    speed=1.0          # Optional: 0.25 to 4.0 (1.0 is normal speed)
)

# Save the audio to a file
output_file = "high_quality_audio.mp3"
response.stream_to_file(output_file)

print(f"✅ Audio generated successfully: {output_file}")

# Play the audio directly in Colab
display(Audio(output_file, autoplay=False))

# Optional: Download the file
from google.colab import files
print("\nDownloading audio file...")
files.download(output_file)
```
